```python
# SECTION D: OUTLIER ANALYSIS
# Q25. Create boxplots for Age and Annual Salary
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.boxplot(data=df_clean, y='Age', ax=axes[0], color='lightgreen')
axes[0].set_title("Q25: Age Boxplot")
sns.boxplot(data=df_clean, y='EstimatedSalary', ax=axes[1], color='coral')
axes[1].set_title("Q25: Salary Boxplot")
plt.savefig("q25_boxplots.png")
plt.show()
plt.close()
```

```python
# Q26–Q28. IQR and Outliers for Age
Q1_age = df_clean['Age'].quantile(0.25)
Q3_age = df_clean['Age'].quantile(0.75)
IQR_age = Q3_age - Q1_age
lower_age = Q1_age - 1.5 * IQR_age
upper_age = Q3_age + 1.5 * IQR_age
age_outliers = df_clean[(df_clean['Age'] < lower_age) | (df_clean['Age'] > upper_age)]

print(f"\n--- Q26-Q28: Age Outlier Analysis ---")
print(f"Age Q1: {Q1_age}, Q3: {Q3_age}, IQR: {IQR_age}")
print(f"Age Bounds: Lower = {lower_age}, Upper = {upper_age}")
print(f"Age Outliers Count: {len(age_outliers)}")
```

```python
# Q29–Q31. IQR and Outliers for Annual Salary
Q1_sal = df_clean['EstimatedSalary'].quantile(0.25)
Q3_sal = df_clean['EstimatedSalary'].quantile(0.75)
IQR_sal = Q3_sal - Q1_sal
lower_sal = Q1_sal - 1.5 * IQR_sal
upper_sal = Q3_sal + 1.5 * IQR_sal
sal_outliers = df_clean[(df_clean['EstimatedSalary'] < lower_sal) | (df_clean['EstimatedSalary'] > upper_sal)]

print(f"\n--- Q29-Q31: Salary Outlier Analysis ---")
print(f"Salary Q1: {Q1_sal}, Q3: {Q3_sal}, IQR: {IQR_sal}")
print(f"Salary Bounds: Lower = {lower_sal}, Upper = {upper_sal}")
print(f"Salary Outliers Count: {len(sal_outliers)}")
```

```python
# Q32. Total number of outliers
total_outliers = len(age_outliers) + len(sal_outliers)
print(f"\n--- Q32: Total Outliers Count ---: {total_outliers}")
```

```python
# Q33. Remove outliers using IQR method
df_no_outliers = df_clean[
    (df_clean['Age'] >= lower_age) & (df_clean['Age'] <= upper_age) &
    (df_clean['EstimatedSalary'] >= lower_sal) & (df_clean['EstimatedSalary'] <= upper_sal)
]
```

```python
# Q34. Compare dataset size before and after outlier removal
print(f"\n--- Q34: Dataset Size Comparison ---")
print(f"Size Before Outlier Removal: {len(df_clean)}")
print(f"Size After Outlier Removal:  {len(df_no_outliers)}")
```

```python
# SECTION F: FEATURE ENGINEERING
df_fe = df_clean.copy()
# Q51. Create Salary_Age_Ratio
df_fe['Salary_Age_Ratio'] = df_fe['EstimatedSalary'] / df_fe['Age']
df_fe.head()
```

```python
# Q52. Significance of Salary_Age_Ratio
print("\n--- Q52: Salary_Age_Ratio Significance ---")
print("Measures earning capability relative to life stage, capturing financial maturity.")
```

```python
# Q53. Create Age Group categories (Young: 18-30, Adult: 31-45, Senior: 46-60)
def assign_age_group(age):
    if 18 <= age <= 30:
        return 'Young'
    elif 31 <= age <= 45:
        return 'Adult'
    elif 46 <= age <= 60:
        return 'Senior'
    return 'Unknown'
df_fe['Age_Group'] = df_fe['Age'].apply(assign_age_group)
df_fe.head()
```

```python
# Q54. Encode Age Group into numerical format
age_map = {'Young': 0, 'Adult': 1, 'Senior': 2}
df_fe['Age_Group_Encoded'] = df_fe['Age_Group'].map(age_map)
df_fe.head()
```

```python
# Q55. Create Salary Categories
def assign_salary_cat(salary):
    if salary < 50000:
        return 'Low Income'
    elif 50000 <= salary <= 100000:
        return 'Middle Income'
    else:
        return 'High Income'
df_fe['Salary_Category'] = df_fe['EstimatedSalary'].apply(assign_salary_cat)
df_fe.head()
```

```python
# Q56. Encode Salary Categories into numerical format
sal_map = {'Low Income': 0, 'Middle Income': 1, 'High Income': 2}
df_fe['Salary_Category_Encoded'] = df_fe['Salary_Category'].map(sal_map)
df_fe.head()
```

```python
# Q57. Create Customer Segment feature
df_fe['Customer_Segment'] = df_fe['Age_Group'] + '_' + df_fe['Salary_Category']
df_fe.head()
```
