```python
# Section A: Data Loading and Dataset Understanding
# Q1. Load dataset & display first 10 records
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('weather_prediction_dataset.csv')
print(df.head(10))
```

```python
# Section C: Data Preprocessing
# Q18. Check missing values in each attribute
print("\nQ18. Missing Values Count:")
print(df.isnull().sum())
```

```python
# Q19. Handle missing values (Mean/Median/Mode imputation if needed)
# Numerical columns imputed with median; Categorical with mode
for col in df.select_dtypes(include=[np.number]).columns:
    df[col] = df[col].fillna(df[col].median())
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Q20. Identify duplicate records
duplicate_count = df.duplicated().sum()
print(f"\nQ20. Duplicate Records Identified: {duplicate_count}")
```

```python

# Q21. Remove duplicate records and verify result
df = df.drop_duplicates()
print(f"Q21. Duplicate Records After Removal: {df.duplicated().sum()}")
```

```python

# Q22. Convert categorical attributes (Weather Condition & RainTomorrow) to numerical
df['Weather_Condition_Encoded'] = df['Weather Condition'].astype('category').cat.codes
df['RainTomorrow_Encoded'] = df['RainTomorrow'].map({'No': 0, 'Yes': 1})

```

```python
# Q23. Verify dataset after preprocessing
print("\nQ23. Updated Dataset Info:")
df.info()
```

```python
# Section D: Outlier Analysis
# Q25. Create boxplots for Temperature and Humidity
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.boxplot(y=df['Temperature'], color='coral')
plt.title("Q25. Temperature Boxplot")
plt.subplot(1, 2, 2)
sns.boxplot(y=df['Humidity'], color='lightgreen')
plt.title("Q25. Humidity Boxplot")
plt.show()
```

```python

# Q26. Calculate Q1, Q3, and IQR for Temperature
q1_temp = df['Temperature'].quantile(0.25)
q3_temp = df['Temperature'].quantile(0.75)
iqr_temp = q3_temp - q1_temp
print(f"\nQ26. Temp Q1: {q1_temp:.2f}, Q3: {q3_temp:.2f}, IQR: {iqr_temp:.2f}")
```

```python

# Q27. Lower and upper bounds for Temperature
lower_temp = q1_temp - 1.5 * iqr_temp
upper_temp = q3_temp + 1.5 * iqr_temp
print(f"Q27. Temp Lower Bound: {lower_temp:.2f}, Upper Bound: {upper_temp:.2f}")
```

```python

# Q28. Identify Temperature outliers
temp_outliers = df[(df['Temperature'] < lower_temp) | (df['Temperature'] > upper_temp)]
print("\nQ28. Temperature Outliers:")
print(temp_outliers[['Date', 'Temperature']])
```

```python

# Q29. Calculate Q1, Q3, and IQR for Humidity
q1_hum = df['Humidity'].quantile(0.25)
q3_hum = df['Humidity'].quantile(0.75)
iqr_hum = q3_hum - q1_hum
print(f"\nQ29. Humidity Q1: {q1_hum:.2f}, Q3: {q3_hum:.2f}, IQR: {iqr_hum:.2f}")
```

```python

# Q30. Lower and upper bounds for Humidity
lower_hum = q1_hum - 1.5 * iqr_hum
upper_hum = q3_hum + 1.5 * iqr_hum
print(f"Q30. Humidity Lower Bound: {lower_hum:.2f}, Upper Bound: {upper_hum:.2f}")

```

```python
# Q31. Identify Humidity outliers
hum_outliers = df[(df['Humidity'] < lower_hum) | (df['Humidity'] > upper_hum)]
print("\nQ31. Humidity Outliers:")
print(hum_outliers[['Date', 'Humidity']])
```

```python

# Q32. Count total number of outliers
total_outliers = len(temp_outliers) + len(hum_outliers)
print(f"\nQ32. Total Outliers Count (Temp + Hum): {total_outliers}")
```

```python

# Q33. Remove outliers using IQR method to create cleaned dataset
df_cleaned = df[
    (df['Temperature'] >= lower_temp) & (df['Temperature'] <= upper_temp) &
    (df['Humidity'] >= lower_hum) & (df['Humidity'] <= upper_hum)
].copy()
```

```python

# Q34. Compare dataset size before and after removal
print(f"\nQ34. Original Dataset Size: {len(df)}, Cleaned Dataset Size: {len(df_cleaned)}")
```

```python

# Q44. Identify weather patterns from scatter plot
# Explanation: Rainy days cluster heavily in regions where Humidity > 60% and Temperature is moderate.

# Q45. Pair plot for all numerical features
num_cols = ['Temperature', 'Humidity', 'Wind Speed', 'Pressure', 'Visibility', 'Cloud Cover', 'Precipitation']
sns.pairplot(df_cleaned[num_cols + ['RainTomorrow']], hue='RainTomorrow')
plt.suptitle("Q45. Pair Plot of Numerical Features", y=1.02)
plt.show()
```

```python

# Q46. Create correlation matrix
corr_matrix = df_cleaned[num_cols + ['RainTomorrow_Encoded']].corr()
print("\nQ46. Correlation Matrix:")
print(corr_matrix)
```

```python

# Q47. Visualize correlation matrix using heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Q47. Correlation Matrix Heatmap")
plt.show()
```

```python

# Q48. Identify attributes most strongly correlated with RainTomorrow
target_corr = corr_matrix['RainTomorrow_Encoded'].abs().sort_values(ascending=False)
print("\nQ48. Absolute Correlations with RainTomorrow:")
print(target_corr)

```

```python

# Q49. Determine whether Temperature or Humidity has stronger influence
temp_corr = abs(corr_matrix.loc['Temperature', 'RainTomorrow_Encoded'])
hum_corr = abs(corr_matrix.loc['Humidity', 'RainTomorrow_Encoded'])
stronger = "Humidity" if hum_corr > temp_corr else "Temperature"
print(f"\nQ49. Stronger Predictor: {stronger} (Hum Corr: {hum_corr:.3f} vs Temp Corr: {temp_corr:.3f})")
```

```python

# Q50. Analyze rainfall patterns across temperature ranges
df_cleaned['Temp_Bin'] = pd.cut(df_cleaned['Temperature'], bins=3)
print("\nQ50. Rain Occurrence Rate by Temperature Bins:")
print(df_cleaned.groupby('Temp_Bin')['RainTomorrow_Encoded'].mean() * 100)
```

```python
# Section F: Feature Engineering
# Q51. Create Temperature_Humidity_Ratio feature
df_cleaned['Temperature_Humidity_Ratio'] = df_cleaned['Temperature'] / df_cleaned['Humidity']
```

```python

# Q52. Explain significance of Temperature_Humidity_Ratio
# Explanation: Captures relative air saturation per degree of heat; low ratios indicate cool,
# moisture-laden air conducive to cloud condensation.

# Q53 & Q54. Create Temperature Categories & Encode
def categorize_temp(t):
    if t < 15:
        return 'Cold'
    elif 15 <= t <= 30:
        return 'Moderate'
    else:
        return 'Hot'

df_cleaned['Temp_Category'] = df_cleaned['Temperature'].apply(categorize_temp)
df_cleaned['Temp_Category_Encoded'] = df_cleaned['Temp_Category'].map({'Cold': 0, 'Moderate': 1, 'Hot': 2})
```

```python

# Q55 & Q56. Create Humidity Categories & Encode
def categorize_humidity(h):
    if h < 50:
        return 'Low'
    elif 50 <= h <= 75:
        return 'Medium'
    else:
        return 'High'

df_cleaned['Humidity_Category'] = df_cleaned['Humidity'].apply(categorize_humidity)
df_cleaned['Humidity_Category_Encoded'] = df_cleaned['Humidity_Category'].map({'Low': 0, 'Medium': 1, 'High': 2})
```

```python

# Q57. Create Weather Risk feature
df_cleaned['Weather_Risk'] = (df_cleaned['Humidity'] * 0.4) + (df_cleaned['Cloud Cover'] * 5) + (df_cleaned['Precipitation'] * 2)
```

```python

# Q58. Visualize distribution of Temperature Categories
plt.figure(figsize=(5, 4))
sns.countplot(x='Temp_Category', data=df_cleaned, order=['Cold', 'Moderate', 'Hot'])
plt.title("Q58. Temperature Categories Distribution")
plt.show()
```

```python

# Q59. Visualize distribution of Humidity Categories
plt.figure(figsize=(5, 4))
sns.countplot(x='Humidity_Category', data=df_cleaned, order=['Low', 'Medium', 'High'])
plt.title("Q59. Humidity Categories Distribution")
plt.show()
```

```python

# Q60. Analyze relationship between Temperature Categories and RainTomorrow
print("\nQ60. Rain Occurrence by Temp Category:")
print(df_cleaned.groupby('Temp_Category')['RainTomorrow_Encoded'].mean() * 100)
```

```python

# Q61. Analyze relationship between Humidity Categories and RainTomorrow
print("\nQ61. Rain Occurrence by Humidity Category:")
print(df_cleaned.groupby('Humidity_Category')['RainTomorrow_Encoded'].mean() * 100)
```
